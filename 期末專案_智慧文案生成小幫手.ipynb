```python
from diffusers import StableDiffusionPipeline, UniPCMultistepScheduler
import torch
import gc
import matplotlib.pyplot as plt
import gradio as gr
import random
```

```python
# 指定圖片模型
model_name = "digiplay/majicMIX_realistic_v6"
```

```python
#注意有可能要停用 use_safetensors=True。

pipe = StableDiffusionPipeline.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    use_safetensors=True
).to("cuda")
```

```python
pipe.scheduler = UniPCMultistepScheduler.from_config(pipe.scheduler.config)
```

```python
from openai import OpenAI
import os
import requests
from PIL import Image
from io import BytesIO
from google.colab import userdata

# Groq 模型設定
import os
api_key = userdata.get('Groq')
os.environ['GROQ_API_KEY']=api_key
client = OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=api_key
)
provider_planner = provider_writer = "groq"
model_planner = model_writer = "gemma2-9b-it"
```

```python
def reply(system="請用台灣習慣的中文回覆。",
          prompt="Hi",
          provider="groq",
          model="llama3-70b-8192"
          ):
    chat_completion = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": prompt}
        ]
    )
    return chat_completion.choices[0].message.content.strip()

# 打造文案(二階段)
system_planner = "請用台灣習慣的中文回應: 你的想法很多元，風格多變，可以根據產品特色、目標客群撰寫文案。"
system_writer = "請用台灣習慣的中文回應: 你是有很多想法的貼文小幫手，最會唬爛了。根據產品的特色或關鍵字，寫出讓人想看的文案。請根據關鍵字加入Hashtag，把它變成一則有創意的社群行銷貼文。"
```

```python
# 生成函式
def generate_images(prompt, use_enhance, enhance_text, use_negative, negative_text,
                    use_custom_seed, custom_seed, height, width, steps, num_images):

    height = int(height)
    width = int(width)

    if height % 8 != 0 or width % 8 != 0:
        raise ValueError("高度和寬度必須是8的倍數！")

    if use_custom_seed:
        base_seed = int(custom_seed)
    else:
        base_seed = random.randint(0, 2**32 - 1)

    seeds = [base_seed + i for i in range(num_images)]

    prompts = []
    negative_prompts = []
    generators = []

    final_prompt = prompt
    if use_enhance and enhance_text:
        final_prompt = prompt + ", " + enhance_text

    final_negative = negative_text if use_negative else None

    for seed in seeds:
        g = torch.Generator("cuda").manual_seed(seed)
        generators.append(g)
        prompts.append(final_prompt)
        negative_prompts.append(final_negative)

    gc.collect()
    torch.cuda.empty_cache()

    images = []
    for i in range(num_images):
        with torch.no_grad():
            image = pipe(
                prompt=prompts[i],
                negative_prompt=negative_prompts[i] if final_negative else None,
                height=height,
                width=width,
                num_inference_steps=steps,
                guidance_scale=7.5,
                generator=generators[i]
            ).images[0]
            images.append(image)

    return images, f"使用的 random seeds: {seeds}"
```

```python
def lucky_post(prompt):
    # Step 1: 規劃理由
    planning_prompt = f"使用者說：{prompt}。請幫我想文案。"
    lucky_reasons = reply(system_planner, planning_prompt, provider_planner, model_planner)

    # Step 2: 從理由中挑選最佳的寫成貼文
    generation_prompt = f"這是我想到的理由：\n{lucky_reasons}\n\n請從中選出一個最有趣的理由，然後根據它寫出一段吸引人的貼文。"
    final_post = reply(system_writer, generation_prompt, provider_writer, model_writer)

    # ----- chatgpt幫助-----
    # Step 3: 翻譯使用者輸入 + 精選理由成英文圖像 prompt
    translate_prompt = f"""請將以下繁體中文轉成圖像生成用的英文提示詞：
    ---
    產品描述：「{prompt}」
    文案理由摘要（擇一精簡描述即可）：「{lucky_reasons.splitlines()[0]}」
    ---
    請輸出英文提示詞，並包含 "masterpiece", "high quality" 類語彙以提高圖像品質。"""
    image_prompt = reply("你是一位文案圖像生成專家", translate_prompt, provider_writer, model_writer)

    # Step 4: 生成圖片
    images, seed_info = generate_images(
        prompt=image_prompt,
        use_enhance=False,
        enhance_text="",
        use_negative=True,
        negative_text="ugly, blurry, bad quality",
        use_custom_seed=False,
        custom_seed=0,
        height=512,
        width=512,
        steps=30,
        num_images=1
    )

    #return lucky_reasons, final_post, images[0]
    return final_post, images[0]
```

```python
# 變為ig模式
import base64
# 輔助函式：PIL 圖片轉 base64（給 HTML 用）
def image_to_base64(image):
    buffered = BytesIO()
    image.save(buffered, format="PNG")
    img_str = base64.b64encode(buffered.getvalue()).decode()
    return img_str

def lucky_post_with_ig_style(prompt):
    final_post, pil_image = lucky_post(prompt)  # 呼叫已有的文案+圖像產生函式

    # 轉成 base64 字串給 img src 用
    img_b64 = image_to_base64(pil_image)

    # IG 貼文風格 HTML
    ig_html = f"""
    <div style="max-width: 400px; border: 1px solid #ddd; border-radius: 12px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, Oxygen, Ubuntu, Cantarell, 'Open Sans', 'Helvetica Neue', sans-serif;">
        <div style="padding: 12px; display: flex; align-items: center;">
            <img src="https://upload.wikimedia.org/wikipedia/commons/9/99/Sample_User_Icon.png"
                 style="width: 44px; height: 44px; border-radius: 50%; margin-right: 12px;">
            <strong style="font-size: 16px;">louisa_tea</strong>
        </div>
        <img src="data:image/png;base64,{img_b64}" style="width: 100%; border-top: 1px solid #eee; border-bottom: 1px solid #eee;">
        <div style="padding: 12px; font-size: 14px; line-height: 1.4; white-space: pre-wrap;">
            {final_post}
        </div>
    </div>
    """

    return ig_html
```

```python
# 文案生成過程
with gr.Blocks() as demo:
    gr.Markdown("### d(d＇∀＇)文案小幫手很高興為您服務!")
    gr.Markdown("請輸入產品描述、目標客群等，讓我幫你變成文案~")

    user_input = gr.Textbox(label="產品介紹（可包含特色、目標客群等）")
    btn = gr.Button("登登登~文案生成d(`･∀･)b")
    btn1 = gr.Button("上傳至Instagram")

    with gr.Tabs():
        with gr.TabItem("📣 生成文案"):
            # out1 = gr.Textbox(label="🧠 創作理由")
            out2 = gr.Textbox(label="📝 最終文案")
            out4 = gr.Image(label="🖼️ 自動配圖")

        with gr.TabItem("📷 IG 模擬貼文"):
            ig_html = gr.HTML()

    # 綁定按鈕的 click 事件
    btn.click(fn=lucky_post, inputs=[user_input], outputs=[out2, out4])
    btn1.click(fn=lucky_post_with_ig_style, inputs=[user_input], outputs=[ig_html])
```
